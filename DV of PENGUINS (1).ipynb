### lineplot

```python
sns.lineplot(y='bill_length_mm',x='bill_depth_mm',data=df,ci=None)#confidence interval
plt.show()
```

```python
sns.lineplot(y='bill_length_mm',x='bill_depth_mm',data=df,hue='species',ci=None)#confidence interval
plt.show()
```

```python
sns.lineplot(y='flipper_length_mm',x='species',data=df,hue='sex',ci=None)
plt.title('Flipper length over species by sex')
plt.show()
```

```python
sns.lineplot(x='body_mass_g',y='flipper_length_mm',data=df,hue='species',ci=None)
plt.title('flipper length trends by body mass')
plt.show()
```

### scatterplot

```python
sns.scatterplot(x='bill_length_mm',y='bill_depth_mm',data=df,hue='species')
plt.title("scatter plot")
plt.show()
```

## Barplot(categorical data)

```python
sns.barplot(x='species',y='body_mass_g',data=df)
plt.title("Average body mass by species")
plt.show()
```

## Count plot (value_count)
##a countplot bascially counts the categorical n returns a count of occuran

```python
sns.countplot(x='island',hue='species',data=df)
plt.title('Penguin Counts per Island')
plt.show()
```

## boxplot (to find outliers)

```python
sns.boxplot(x='species',y='bill_length_mm',data=df)
plt.title('bill length distribution by species')
plt.show()
```

```python
sns.boxplot(x='species',y='bill_depth_mm',data=df)
plt.title('bill depth distribution by species')
plt.show()
```

```python
sns.boxplot(x='species',y='flipper_length_mm',data=df)
plt.title('Flipper length distribution by species')
plt.show()
```

### Violinplot (density of data)

```python
sns.violinplot(x='species',y='body_mass_g',hue='sex',data=df)
plt.title('body mass distribution by species & sex')
plt.show()
```

```python
sns.violinplot(x='species',y='bill_length_mm',hue='sex',data=df)
plt.title('bill length distribution by species & sex')
plt.show()
```

```python
sns.violinplot(x='species',y='flipper_length_mm',hue='sex',data=df)
plt.title('flipper length distribution by species & sex')
plt.show()
```

## stripplot
single numerical axis typically grp by catagorical variable

```python
sns.stripplot(x='species',y='flipper_length_mm',hue='sex',data=df)
plt.title('flipper length distribution by species n sex')
plt.show()
```

## Histogram = contiouns data n distribution

```python
sns.histplot(x='body_mass_g',data=df,hue='sex',kde=True)
plt.title('distribution of body mass ')
plt.show()
```

```python
sns.histplot(x='flipper_length_mm',kde=True,hue='species',data=df)
plt.title('flipper length over species')
plt.show()
```

## distplot -- univariant set of observation n visualization i.e like histogram


```python
sns.displot(data=df,x='flipper_length_mm',color='g',kde=True)
plt.show()
```

## pairplot

```python
sns.pairplot(df,hue='species')
plt.show()
```

### Catagorical

```python
sns.catplot(x='species',y='flipper_length_mm',data=df,hue='sex',kind='box')
plt.title("flipper length distribution")
plt.show()
```

```python
sns.catplot(x='species',y='bill_length_mm',data=df,hue='sex',kind='box')
plt.title("bill length distribution")
plt.show()
```

```python
sns.catplot(x='species',y='bill_length_mm',data=df,hue='sex',kind='violin')
plt.title("bill length distribution")
plt.show()
```
