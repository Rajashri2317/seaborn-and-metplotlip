### Datavisualization of healthexp

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset("healthexp")
```

## lineplot

```python
sns.lineplot(x='Year',y='Life_Expectancy',data=df,ci=None)#confidence interval
plt.show()
```

## scatterplot

```python
sns.scatterplot(x='Year',y='Life_Expectancy',data=df,hue='Country')
plt.title("scatter plot")
plt.show()
```

## Barplot

```python
sns.barplot(x='Country',y='Life_Expectancy',data=df)
plt.show()
```

## countplot

```python
sns.countplot(x='Country',data=df)
plt.show()
```

#  Boxplot

```python
sns.boxplot(x='Life_Expectancy',data=df)
plt.show()
```

```python
sns.boxplot(x='Spending_USD',data=df)
plt.show()
```

## Violinplot

```python
sns.violinplot(x='Country',y='Life_Expectancy',data=df)
plt.show()
```

```python
sns.violinplot(x='Country',y='Spending_USD',data=df)
plt.show()
```

## stripplot

```python
sns.stripplot(x='Country',y='Life_Expectancy',data=df)
plt.show()
```

```python
sns.stripplot(x='Country',y='Spending_USD',data=df)
plt.show()
```

## Histogram

```python
sns.histplot(x='Life_Expectancy',data=df)
plt.show()
```

```python
sns.histplot(x='Spending_USD',data=df)
plt.show()
```

```python
sns.histplot(x='Year',data=df)
plt.show()
```

```python
sns.histplot(x='Country',data=df)
plt.show()
```

## catagorical 

```python
sns.catplot(x='Life_Expectancy',data=df,kind='violin')
plt.show()
```

```python
sns.catplot(x='Spending_USD',data=df,kind='box')
plt.show()
```

```python
sns.catplot(y='Life_Expectancy',x='Country',data=df,kind='swarm') #striplot
plt.show()
```
