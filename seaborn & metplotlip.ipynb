```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# getting the built-in datasets of seaborn
sns.get_dataset_names()
```

```python
#iris is kind of flower
#3 types == setosa,versicolor,virinica 
#based on l,b of petals n stepls
```

```python
df=sns.load_dataset('iris')
```

## lineplot

```python
sns.lineplot(x='sepal_length',y='sepal_width',data=df,ci=None)#confidence interval
plt.show()
```

```python
### line plot
sns.lineplot(x='sepal_length',y='sepal_width',data=df,hue='species',ci=None)
plt.show()
```

```python
sns.lineplot(x='petal_length',y='species',data=df,hue='species',ci=None)
plt.show()
```

## Scatter plot

```python
##scatter plot
sns.scatterplot(x='sepal_length',y='sepal_width',data=df,hue='species')
plt.title("scatter plot")
plt.show()
```

```python
sns.scatterplot(x='petal_length',y='petal_width',data=df,hue='species')
plt.title('Iris Flower Species by Petal Dimensions')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```

## pairplot

```python
sns.pairplot(df,hue='species')
plt.show()
```

## barplot (estimate for comparing categorical data)

```python

#it is used to aggregate the c.d according to some methods)
sns.barplot(x='species',y='sepal_length',data=df)
plt.show()
```

### count plot value_count()
##a countplot bascially  counts the categorical n returns a count of occuran

```python
sns.countplot(x='species',data=df)
plt.show
```

## Violinplot

```python
##density of data
sns.violinplot(x='species',y='sepal_length',data=df)
plt.show()
```

```python
sns.violinplot(x='species',y='sepal_width',data=df)
plt.show()
```

## stripplot

```python
sns.stripplot(x='species',y='petal_width',data=df)
plt.show()
```

## Histogram = contiouns data n distribution

```python
sns.histplot(x='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x='petal_width', hue='species',data=df,kde=True) #kernal density estimation is a appox of histogram density curve.
plt.show()
```

### distplot -- univariant set of observation n visualization i.e like histogram

```python
sns.displot(data=df,x='petal_width',color='g',kde=True)
plt.show()
```

### Catagorical

```python
sns.catplot(x='sepal_length',data=df,kind='violin')
plt.show()
```

```python
sns.catplot(y='sepal_length',x='species',data=df,kind='swarm') #striplot
plt.show()
```

```python
sns.catplot(y='petal_length',x='species',data=df,kind='swarm') #striplot
plt.show()
```
